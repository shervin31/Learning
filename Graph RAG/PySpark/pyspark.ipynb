```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('Practice').getOrCreate()
```

```python
df_pyspark = spark.read.csv("bank data.csv")
df_pyspark.show(5)
```

```python
## Spark treats it like a normal csv file, so you must specify to get rid of the c's.
df_pyspark = spark.read.option('header', 'true').csv("bank data.csv", inferSchema=True)     ## inferSchema preserves real datatypes of variables
df_pyspark.show(5)
```

```python
## Show all variables
df_pyspark.printSchema()
```
