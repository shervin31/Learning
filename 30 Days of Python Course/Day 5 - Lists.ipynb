### List Methods 

```python
# Checking Items 
lst = ['banana', 'apple', 'pear', 'orange']
condition = 'apple' in lst
print(condition)

# .insert(index, item) inserts item into certain index 
lst.insert(1, 'pineapple')
print(lst)

# .remove(item) removes certain item from list 
lst.remove('orange')
print(lst)

# .pop(index) removes at an index
lst.pop(0)
print(lst)

# del keyword removes at an index or within a range
del lst[0: 2]
print(lst)

# .count() counts number of times item appears in list 
lst = ['banana', 'apple', 'pear', 'orange', 'orange']
print(lst.count('orange'))

# .sort(reverse = True/False) sorts list 
lst.sort(reverse=False)
print(lst)
```
