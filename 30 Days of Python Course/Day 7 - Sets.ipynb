# Sets 
- Sets are a collection of unique values

```python
st = set({'banana', 'orange', 'mango', 'lemon'})
print(st)
```

### String Methods

```python
# .add() adds items 
st.add('apple')
print(st)

# .update([list]) adds multiple items to a set. Takes List as parameter
st.update(['peach', 'lime', 'pear'])
print(st)

# .remove() removes items from set 
st.remove('pear')
print(st)

# Converting list to set 
lst = ['item1', 'item2', 'item3', 'item4', 'item1']
s = set(lst)
print(s)

# .intersection() finds intersection of 2 sets
st1 = {'item1', 'item2', 'item3', 'item4'}
st2 = {'item3', 'item2'}
print(st1.intersection(st2))

# .issubset()/.issuperset checks if it is a subet/superset
print(st2.issubset(st1))
print(st2.issuperset(st2))
```
