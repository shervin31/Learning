# Dicionary

- Dictionary is a collection of unordered mutable paired (key : value) data type

- To create a dictionary we use curly brackets, {} or the dict() built-in function.

 


```python
dct = {
    'first_name': 'Shervin',
    'last_name': 'Zare',
    'Age': 20, 
    'Country': 'Iran',
    'is_married' : 'No', 
    'Skills': ['python', 'SQL', 'C', 'Java']
}

# Accessing Dictionary Items 
print(dct['Age'])
print('Java' in dct['Skills'])
```

### Dictionary Methods 


```python
# Adding key pair value 
dct['Height'] = '170cm'

# Modifying values 
dct['Skills'].append('R')

# .pop(key) removes and returns item with specified key
dct.pop('is_married')

# .popitem() removes and returns last item 
dct.popitem()
dct

# del just deletes item with specified key 
del dct['Age']
dct

# .keys() and .values() 
print("Here are the keys: {}".format(dct.keys()))
print("Here are the values: {}".format(dct.values()))
```
