# Importing Built-in Modules 

### OS Module (Operating System) 

```python
import os 

# Create a directory in current directory (folder in current folder)
os.mkdir('projects')

# Changing the current directory 
os.chdir('path')

# Get current directory
os.getcwd()

# Remove directory 
os.rmdir()
```

### Statistics Module

```python
from statistics import *

ages = [20, 20, 4, 24, 25, 22, 26, 20, 23, 22, 26]
print(mean(ages))
print(median(ages))
print(mode(ages))
```

### Random Module

```python
from random import random, randint

# Returns random value between 0 and .9999
print(random())

# Returns random integer between range
print(randint(5,29))
```
